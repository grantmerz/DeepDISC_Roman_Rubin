# Resampling ablation: magnitude-binned curves (paper Appendix C)

Magnitude-binned metrics for the four Rubin-only models of the resampling ablation (`tab:ups_two_by_two`), each at
its own F1-optimal operating point from the validation grid search. The same point is the F1 maximum at 1.0″, the
F1 maximum at 0.5″, and first in `select_outs/rankings_mean_ll.csv`:

| Model | Run | (θ_ROI, θ_NMS) |
|---|---|---|
| DeepDISC Rubin | `lsst5_30k_4h200_bs192_ep50` | (0.50, 0.60) |
| DeepDISC RubinUps | `lsst5_ups_30k_4h200_bs192_ep50` | (0.60, 0.65) |
| DeepDISC PatchSize13 | `lsst5_ps13_ups_30k_4h200_bs192_ep50` | (0.50, 0.60) |
| DeepDISC RubinPadded | `lsst5_pad_30k_4h200_bs192_ep50` | (0.50, 0.60) |

Same settings as the "Paper Runs" cell in `metrics.ipynb`. `plot_binned_metrics` colours runs by their order in
`run_pairs`, so this order keeps Rubin and RubinUps in their main-paper colours and gives PS13 the purple it has in the
Appendix D anchor figures. The padded model gets grey.

```python
%load_ext autoreload
%autoreload 2
from pathlib import Path
from IPython.display import Image, display
from plot_binned_metrics import main, NotebookArgs
```

```python
RUNS = {  # run -> (operating point, legend label); order sets the colours
    'lsst5_30k_4h200_bs192_ep50':          ('s0.5_n0.6',  'DeepDISC Rubin'),
    'lsst5_ups_30k_4h200_bs192_ep50':      ('s0.6_n0.65', 'DeepDISC RubinUps'),
    'lsst5_ps13_ups_30k_4h200_bs192_ep50': ('s0.5_n0.6',  'DeepDISC PatchSize13'),
    'lsst5_pad_30k_4h200_bs192_ep50':      ('s0.5_n0.6',  'DeepDISC RubinPadded'),
}
args = NotebookArgs(
    run_pairs=[list(RUNS)],
    mapped_combo_pairs=[tuple(c for c, _ in RUNS.values())],
    score_thresholds=[],
    nms_thresholds=[],
    linking_lengths=('0.5', '1.0'),
    dynamic_ylim=True,
    min_count=10,
    run_labels={r: lab for r, (_, lab) in RUNS.items()},
    hide_combo_in_label=True,
    lsst_in_comparison=True,
    plot_mag_min=19,
    lsst_mode='overlay_only',
)
main(args)
```
