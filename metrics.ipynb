## Dominance Check: DD vs LSST in Detection Space

```python
import pandas as pd
from pathlib import Path

# Load grid search CSV for the baseline run
csv_path = Path('/u/yse2/lsst_runs/lsst5_30k_4h200_bs192_ep50/metrics/gold_buf2/gs_metrics.csv').expanduser()
df = pd.read_csv(csv_path)

ll = '1.0'
comp_col = f'dd_completeness_{ll}'
purity_col = f'dd_purity_{ll}'
unrec_blend_col = f'dd_unrec_blend_frac_{ll}'
lsst_comp_col = f'lsst_completeness_{ll}'
lsst_purity_col = f'lsst_purity_{ll}'
lsst_unrec_blend_col = f'lsst_unrec_blend_frac_{ll}'

# Extract LSST's values (should be same for all rows, take first non-null)
lsst_completeness = df[lsst_comp_col].dropna().iloc[0]
lsst_purity = df[lsst_purity_col].dropna().iloc[0]

if comp_col in df.columns and purity_col in df.columns:
    mask = (df[comp_col] >= lsst_completeness) & (df[purity_col] >= lsst_purity)
    count = mask.sum()
    
    print(f"LSST thresholds (LL={ll}\"):")
    print(f"  completeness = {lsst_completeness:.4f}")
    print(f"  purity = {lsst_purity:.4f}\n")
    print(f"Dominating combos: {count}\n")
    
    if count > 0:
        print("✓ DD DOMINATES LSST in detection space")
        print(f"  {count} combo(s) achieve both thresholds simultaneously\n")
        print("Combos that dominate:")
        result = df[mask][['score_thresh', 'nms_thresh', comp_col, purity_col, unrec_blend_col]].copy()
        result.columns = ['score_thresh', 'nms_thresh', 'completeness', 'purity', 'unrec_blend_frac']
        print(result.to_string(index=False))
    else:
        print("✗ LSST sits on DD's Pareto frontier")
        print("  No combos exceed both thresholds simultaneously")
        print("  DD can move along the frontier but cannot improve both metrics together")
else:
    print(f"Columns not found: {comp_col}, {purity_col}")
```

```python
import pandas as pd
from select_thresh import (
    pareto_set, percentile, select_operating_point,
    mean_percentile_across_lls, stability, col,
)
# comb_30k_4h200_bs144_ep50
# 
df = pd.read_csv("/u/yse2/lsst_runs/comb_30k_4h200_bs144_ep50/metrics/gold_buf2/gs_metrics.csv")
P, LL, SET = "dd", "1.0", ["f1", "unrec_blend_frac"]

i_star = df[col(P, "f1", LL)].idxmax()
star   = df.loc[i_star]
combo  = (star["score_thresh"], star["nms_thresh"])
print(f"max-F1 combo: s{combo[0]}_n{combo[1]}")

# A: UBF (and F1) percentile of max-F1 combo
ubf_pct, f1_pct = percentile(df, P, "unrec_blend_frac", LL), percentile(df, P, "f1", LL)
print(f"A) F1 pct={f1_pct.loc[i_star]:.1f}  UBF pct={ubf_pct.loc[i_star]:.1f}")

# B: is max-F1 in the (F1,UBF) Pareto set?
mask = pareto_set(df, P, SET, LL)
print(f"B) in Pareto set: {bool(mask[df.index.get_loc(i_star)])}  (set size={mask.sum()})")

# C: percentile-rank over {F1,UBF} -> same/adjacent combo?
op = select_operating_point(df, P, SET, LL)
mean_combo = op.index[0]
min_combo  = op.sort_values("min_pct", ascending=False).index[0]
adj = lambda a, b, s=0.05, e=1e-9: abs(a[0]-b[0]) <= s+e and abs(a[1]-b[1]) <= s+e
print(f"C) mean_pct best: s{mean_combo[0]}_n{mean_combo[1]}  same={mean_combo==combo} adj={adj(mean_combo,combo)}")
print(f"   min_pct  best: s{min_combo[0]}_n{min_combo[1]}  same={min_combo==combo} adj={adj(min_combo,combo)}")

# D: margin over LSST at max-F1 combo
for m in SET:
    lsst = df[col("lsst", m, LL)].dropna().iloc[0]
    print(f"D) {m}: dd={star[col(P,m,LL)]:.4f}  lsst={lsst:.4f}  delta={star[col(P,m,LL)]-lsst:+.4f}")

# E: combos beating LSST on BOTH; is max-F1 one?
lf1, lubf = (df[col('lsst','f1',LL)].dropna().iloc[0],
             df[col('lsst','unrec_blend_frac',LL)].dropna().iloc[0])
beats = (df[col(P,'f1',LL)] > lf1) & (df[col(P,'unrec_blend_frac',LL)] < lubf)
print(f"E) beat LSST on both: {beats.sum()}/{len(df)}  max-F1 among them={bool(beats.loc[i_star])}")

# F: relative F1 span (3x3)
_, span, _ = stability(df, P, "f1", LL, combo)
print(f"F) F1 span={span:.4f}  relative={span/df[col(P,'f1',LL)].max():.4%}")

# G: F1 mean-percentile rank across LLs
rank_ll = mean_percentile_across_lls(df, P, "f1", [0.5, 1.0])
print(f"G) max-F1(1.0) combo rank across LLs: #{list(rank_ll.index).index(combo)+1}/{len(rank_ll)}")

pct_combo = select_operating_point(df, P, SET, LL).index[0]      # mean_pct winner

def rowfor(combo):
    return df[(df.score_thresh==combo[0]) & (df.nms_thresh==combo[1])].iloc[0]

metrics = ["f1","unrec_blend_frac","purity","completeness","resolved_frac","spurious_frac","shred_frac"]
higher_better = {"f1","purity","completeness","resolved_frac"}
for name, combo in [("F1 pick", combo), ("pct winner", pct_combo)]:
    r = rowfor(combo)
    print(f"\n{name}  s{combo[0]}_n{combo[1]}")
    for m in metrics:
        dd, lsst = r[col(P,m,LL)], df[col('lsst',m,LL)].dropna().iloc[0]
        d = dd - lsst
        worse = (d < 0) if m in higher_better else (d > 0)
        print(f"  {m:18s} dd={dd:.4f} lsst={lsst:.4f} Δ={d:+.4f}{'  <-- worse than LSST' if worse else ''}")
```

```python
# === check_operating_point & compare_picks ===
# Insert this cell near the top (after imports), or keep it here and run it before the analysis cells.
# Replace the fragment in Cell 29 with: check_operating_point(run="comb30k", LL="1.0")

from select_thresh import (
    pareto_set, percentile, select_operating_point,
    mean_percentile_across_lls, stability, col,
)
import pandas as pd
import numpy as np

_MODEL_DIRS = {
    "lsst30k":  "lsst5_30k_4h200_bs192_ep50",
    "ups30k":   "lsst5_ups_30k_4h200_bs192_ep50",
    "comb30k":  "comb_30k_4h200_bs144_ep50",
    # passthrough for raw run names
    "lsst5_30k_4h200_bs192_ep50":      "lsst5_30k_4h200_bs192_ep50",
    "lsst5_ups_30k_4h200_bs192_ep50":  "lsst5_ups_30k_4h200_bs192_ep50",
    "comb_30k_4h200_bs144_ep50":       "comb_30k_4h200_bs144_ep50",
}

def check_operating_point(run="lsst30k", mag="gold", buf=2, LL="1.0",
                          lls_for_rank=(0.5, 1.0), pipeline="dd"):
    """Run the full operating-point checks for one model.

    Parameters
    ----------
    run : str
        Short alias ("lsst30k", "ups30k", "comb30k") or full run directory name.
    mag : str
        "gold" or "power_law".
    buf : int
        Buffer (Δm).  2 → gold_buf2, 0 → gold_buf0, etc.
    LL : str
        Primary linking length for per-metric checks (default "1.0").
    lls_for_rank : tuple
        Linking lengths for the cross-LL ranking (default (0.5, 1.0)).
    pipeline : str
        "dd" or "lsst".

    Returns
    -------
    dict with keys: combo, f1_pct, ubf_pct, in_pareto, pareto_size,
                    mean_combo, min_combo, deltas, beats_sum, beats_total,
                    beats_contains, f1_span, f1_rel_span, rank_across_lls,
                    df, star_row
    """
    run_dir = _MODEL_DIRS.get(run, run)
    pfx = "gold" if mag == "gold" else "pl"
    csv_path = f"/u/yse2/lsst_runs/{run_dir}/metrics/{pfx}_buf{buf}/gs_metrics.csv"

    df = pd.read_csv(csv_path)
    P = pipeline
    SET = ["f1", "unrec_blend_frac"]

    i_star = df[col(P, "f1", LL)].idxmax()
    star = df.loc[i_star]
    combo = (star["score_thresh"], star["nms_thresh"])
    print(f"--- {run_dir} ---")
    print(f"max-F1 combo: s{combo[0]}_n{combo[1]}\n")

    # A: UBF (and F1) percentile of max-F1 combo
    ubf_pct_s, f1_pct_s = percentile(df, P, "unrec_blend_frac", LL), percentile(df, P, "f1", LL)
    f1_pct_val = f1_pct_s.loc[i_star]
    ubf_pct_val = ubf_pct_s.loc[i_star]
    print(f"A) F1 pct={f1_pct_val:.1f}  UBF pct={ubf_pct_val:.1f}")

    # B: in (F1, UBF) Pareto set?
    mask = pareto_set(df, P, SET, LL)
    in_pareto = bool(mask[df.index.get_loc(i_star)])
    pareto_size = int(mask.sum())
    print(f"B) in Pareto set: {in_pareto}  (set size={pareto_size})")
    # B2: print full Pareto front members (for reference)
    front = df.iloc[np.flatnonzero(mask)][["score_thresh", "nms_thresh"]]
    print(f"    front: {', '.join(f's{r.score_thresh}_n{r.nms_thresh}' for _, r in front.iterrows())}")
    print(f"    s0.4_n0.65 in front: {((front.score_thresh==0.4) & (front.nms_thresh==0.65)).any()}")
    # C: percentile-rank over {F1,UBF}
    op = select_operating_point(df, P, SET, LL)
    mean_combo = op.index[0]
    min_combo = op.sort_values("min_pct", ascending=False).index[0]
    adj = lambda a, b, s=0.05, e=1e-9: abs(a[0] - b[0]) <= s + e and abs(a[1] - b[1]) <= s + e
    print(f"C) mean_pct best: s{mean_combo[0]}_n{mean_combo[1]}  same={mean_combo==combo} adj={adj(mean_combo, combo)}")
    print(f"   min_pct  best: s{min_combo[0]}_n{min_combo[1]}  same={min_combo==combo} adj={adj(min_combo, combo)}")

    # D: margin over LSST
    deltas = {}
    for m in SET:
        lsst = df[col("lsst", m, LL)].dropna().iloc[0]
        dd = star[col(P, m, LL)]
        d = dd - lsst
        deltas[m] = d
        print(f"D) {m}: dd={dd:.4f}  lsst={lsst:.4f}  delta={d:+.4f}")

    # E: combos beating LSST on both
    lsst_row = df[df[col("lsst", "f1", LL)].notna()].iloc[0]
    lf1 = lsst_row[col("lsst", "f1", LL)]
    lubf = lsst_row[col("lsst", "unrec_blend_frac", LL)]
    beats = (df[col(P, "f1", LL)] > lf1) & (df[col(P, "unrec_blend_frac", LL)] < lubf)
    beats_sum = int(beats.sum())
    beats_total = len(df)
    beats_contains = bool(beats.loc[i_star])
    print(f"E) beat LSST on both: {beats_sum}/{beats_total}  max-F1 among them={beats_contains}")
    # E2: which combos fail to beat LSST on both?
    fail = df[~beats][["score_thresh", "nms_thresh"]]
    print(f"    failing combos by score_thresh:")
    print(fail.groupby("score_thresh").nms_thresh.agg(["min", "max", "count"]).to_string())
    print(pd.crosstab(df.score_thresh, df.nms_thresh, values=beats, aggfunc="first").replace({True:"·", False:"X"}))
    # E3: combos with purity ≥ LSST purity — how close on completeness?
    lsst_purity = df[col("lsst", "purity", LL)].dropna().iloc[0]
    hp = df[df[col(P, "purity", LL)] >= lsst_purity]
    print(f"    combos with purity ≥ LSST ({lsst_purity:.4f}):")
    print(hp[["score_thresh", "nms_thresh", col(P, "purity", LL), col(P, "completeness", LL), col(P, "f1", LL)]].sort_values(col(P, "completeness", LL), ascending=False).head().to_string(index=False))
    # E4: combos dominating LSST on ALL FOUR metrics (purity, completeness, f1, UBF)
    lsst_purity = df[col("lsst", "purity", LL)].dropna().iloc[0]
    lsst_comp = df[col("lsst", "completeness", LL)].dropna().iloc[0]
    dom = ((df[col(P, "purity", LL)] > lsst_purity)
            & (df[col(P, "completeness", LL)] > lsst_comp)
            & (df[col(P, "f1", LL)] > lf1)
            & (df[col(P, "unrec_blend_frac", LL)] < lubf))
    print(f"    dominate LSST on all 4 metrics:{dom.sum()}/{len(df)}")
    if dom.sum() > 0:
        print(df[dom][["score_thresh","nms_thresh"]].to_string(index=False))
        print(f"    max F1 in dom set: {df[dom][col(P, 'f1', LL)].max():.4f}", f"(global max: {df[col(P, 'f1', LL)].max():.4f})")
    # F: relative F1 span (3x3)
    _, span, _ = stability(df, P, "f1", LL, combo)
    f1_rel_span = span / df[col(P, "f1", LL)].max()
    print(f"F) F1 span={span:.4f}  relative={f1_rel_span:.4%}")
    # H: 3×3 neighborhood vs LSST margin (¶7 stability claim)
    s, n = combo
    nb = df[(df["score_thresh"].between(s-0.05, s+0.05))
            & (df["nms_thresh"].between(n-0.05, n+0.05))]
    min_f1_nb = nb[col(P, "f1", LL)].min()
    max_f1_nb = nb[col(P, "f1", LL)].max()
    span_nb = max_f1_nb - min_f1_nb
    lsst_f1 = df[col("lsst", "f1", LL)].dropna().iloc[0]
    lsst_gap = max_f1_nb - lsst_f1
    lsst_margin = min_f1_nb - lsst_f1
    print(f"H) 3×3 neighborhood: F1 ∈ [{min_f1_nb:.4f}, {max_f1_nb:.4f}]  "
            f"span={span_nb:.4f}  LSST gap={lsst_gap:.4f}  "
            f"span/gap={span_nb/lsst_gap:.1%}  "
            f"min−lsst={lsst_margin:.4f}")
    
    # G: F1 mean-percentile rank across LLs
    rank_ll = mean_percentile_across_lls(df, P, "f1", lls_for_rank)
    rank_pos = list(rank_ll.index).index(combo) + 1
    print(f"G) max-F1({LL}\") combo rank across LLs: #{rank_pos}/{len(rank_ll)}")

    return {
        "combo": combo, "f1_pct": f1_pct_val, "ubf_pct": ubf_pct_val,
        "in_pareto": in_pareto, "pareto_size": pareto_size,
        "mean_combo": mean_combo, "min_combo": min_combo,
        "deltas": deltas,
        "beats_sum": beats_sum, "beats_total": beats_total,
        "beats_contains": beats_contains,
        "f1_span": span, "f1_rel_span": f1_rel_span,
        "rank_across_lls": (rank_pos, len(rank_ll)),
        "df": df, "star_row": star,
    }


def compare_picks(run="lsst30k", mag="gold", buf=2, LL="1.0", pipeline="dd"):
    """Side-by-side: F1-optimal combo vs multi-metric percentile winner."""
    run_dir = _MODEL_DIRS.get(run, run)
    pfx = "gold" if mag == "gold" else "pl"
    csv_path = f"/u/yse2/lsst_runs/{run_dir}/metrics/{pfx}_buf{buf}/gs_metrics.csv"
    df = pd.read_csv(csv_path)
    P = pipeline
    SET = ["f1", "unrec_blend_frac"]

    i_star = df[col(P, "f1", LL)].idxmax()
    combo = (df.loc[i_star, "score_thresh"], df.loc[i_star, "nms_thresh"])
    pct_combo = select_operating_point(df, P, SET, LL).index[0]

    def _row(combo):
        return df[(df.score_thresh == combo[0]) & (df.nms_thresh == combo[1])].iloc[0]

    metrics = ["f1", "unrec_blend_frac", "purity", "completeness",
               "resolved_frac", "spurious_frac", "shred_frac"]
    higher_better = {"f1", "purity", "completeness", "resolved_frac"}

    print(f"--- {run_dir} ---")
    for name, c in [("F1 pick", combo), ("pct winner", pct_combo)]:
        r = _row(c)
        print(f"\n{name}  s{c[0]}_n{c[1]}")
        for m in metrics:
            dd = r[col(P, m, LL)]
            lsst = df[col("lsst", m, LL)].dropna().iloc[0]
            d = dd - lsst
            worse = (d < 0) if m in higher_better else (d > 0)
            flag = "  <-- worse than LSST" if worse else ""
            print(f"  {m:18s} dd={dd:.4f} lsst={lsst:.4f} Δ={d:+.4f}{flag}")
```

## Trade-off Plot: Best F1 (star) vs Best Completeness (triangle)

```python
from select_thresh import plot_tradeoff, col
import pandas as pd
from pathlib import Path
%matplotlib inline

# [lsst30k]="$HOME/lsst_runs/lsst5_30k_4h200_bs192_ep50/metrics"
# [ups30k]="$HOME/lsst_runs/lsst5_ups_30k_4h200_bs192_ep50/metrics"
# [comb30k]="$HOME/lsst_runs/comb_30k_4h200_bs144_ep50/metrics"
# [clip30k]="$HOME/lsst_runs/clip5_30k_4h200_bs64_ep50/metrics"
# [lsstall]="$HOME/lsst_runs/lsst5_all_4h200_bs192_ep20/metrics"
base_path = Path('/u/yse2/lsst_runs/')
runs = {'baseline': 'lsst5_30k_4h200_bs192_ep50', 
        'ups': 'lsst5_ups_30k_4h200_bs192_ep50',
        'comb': 'comb_30k_4h200_bs144_ep50'}
ll = '1.0'
def get_gs_path(run_name):
    return base_path / runs[run_name] / 'metrics' / 'gold_buf2' / 'gs_metrics.csv'
def get_out_path(run_name):
    return base_path / runs[run_name] / 'metrics' / 'select_outs'

def plot_custom_tradeoff(run_name):
    df = pd.read_csv(get_gs_path(run_name))
    # mark_best_xy=False so mark_combos alone decides all three markers:
    # star = Best F1 (not an axis here), D = Best Purity, triangle = Best Completeness.
    # Legend marker size is 120 for the star, 75 for D/triangle (set inside plot_tradeoff).
    f1_row = df.loc[df[col('dd', 'f1', ll)].idxmax()]
    f1_combo = (f1_row['score_thresh'], f1_row['nms_thresh'])
    purity_row = df.loc[df[col('dd', 'purity', ll)].idxmax()]
    purity_combo = (purity_row['score_thresh'], purity_row['nms_thresh'])
    completeness_row = df.loc[df[col('dd', 'completeness', ll)].idxmax()]
    completeness_combo = (completeness_row['score_thresh'], completeness_row['nms_thresh'])
    print(f'Best F1: {f1_combo}')
    print(f'Best Purity: {purity_combo}')
    print(f'Best Completeness: {completeness_combo}')
    # Same mask as the Dominance Check cell above -- the 5 combos that beat LSST on
    # both completeness and purity simultaneously. Ringed as one group (not one
    # mark_combos entry each) so the legend doesn't balloon to 8 entries.
    lsst_completeness = df[f'lsst_completeness_{ll}'].dropna().iloc[0]
    lsst_purity = df[f'lsst_purity_{ll}'].dropna().iloc[0]
    # dominates_lsst = ((df[col('dd', 'completeness', ll)] >= lsst_completeness)
    #                 & (df[col('dd', 'purity', ll)] >= lsst_purity))
    fig, ax = plot_tradeoff(
        df, 'dd', x_metric='completeness', y_metric='purity',
        color_metric='score_thresh', ll=ll,
        mark_best_xy=False,
        mark_combos=[
            (f1_combo, '*', 'Best F1'),
            (completeness_combo, '^', 'Best Completeness'),
            (purity_combo, 'D', 'Best Purity'),
        ],
        title=rf'Purity vs Completeness (DD Rubin, LL = {ll}", $\Delta$m = 2)',
        # legend_ncol=2,
        pareto=True
    )
    fig.savefig(get_out_path(run_name) / 'purity_completeness_score_thresh.png', dpi=200, bbox_inches='tight')
    print('Tradeoff plot')
    fig, ax = plot_tradeoff(
        df, 'dd', x_metric='completeness', y_metric='purity',
        color_metric='nms_thresh', ll=ll,
        mark_best_xy=False,
        mark_combos=[
            (f1_combo, '*', 'Best F1'),
            (completeness_combo, '^', 'Best Completeness'),
            (purity_combo, 'D', 'Best Purity'),
        ],
        title=rf'Purity vs Completeness (DD Rubin, LL = {ll}", $\Delta$m = 2)',
            # legend_ncol=2,
            pareto=True
    )
    fig.savefig(get_out_path('baseline') / 'purity_completeness_nms_thresh.png', dpi=200, bbox_inches='tight')
```
